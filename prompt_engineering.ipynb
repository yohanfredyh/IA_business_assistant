# Partie 2  
**Comparer les techniques de prompting**

**technique zero_shot**:

prompt: Classe le commentaire suivant : "Le service est rapide mais l'application plante régulièrement."
Classes possibles : positif, négatif, neutre.

Réponse_LLM: Ce commentaire est mixte, mais si je dois choisir une seule classe parmi les trois proposées, je le classerais en négatif.

**technique one_shot** 

prompt:

Exemple :
Commentaire : "Livraison en retard mais produit conforme."

Classe : neutre

Classe maintenant ce Commentaire : "Le service est rapide mais l'application plante régulièrement."

Classes possibles : positif, négatif, neutre.

Réponse_LLM : Neutre

En suivant l'exemple donné (un commentaire mixte positif/négatif est classé "neutre"), le même raisonnement s'applique ici : "rapide" (positif) + "plante régulièrement" (négatif) = un commentaire ambivalent, donc neutre.


**la technique few-shot**

prompt: 

Exemples :

Commentaire : "Excellent produit, je recommande !" → positif

Commentaire : "Aucun problème particulier, correct." → neutre

Commentaire : "Très déçu, produit défectueux." → négatif

Classe maintenant :

Commentaire : "Le service est rapide mais l'application plante régulièrement."

Classes possibles : positif, négatif, neutre.

Réponse_LLM : Négatif

**prompt structuré**
Tu es un système de classification d'avis clients. Tu dois Classer le commentaire suivant dans une des classes autorisées.

Commentaire : "Le service est rapide mais l'application plante régulièrement."

Classes autorisées : positif, négatif, neutre.

Format de réponse : uniquement le nom de la classe, sans justification.

Réponse_LLM : Négatif

# Partie 6 – Prompt Engineering pour le Machine Learning

1) Créer un prompt permettant, à partir du dataset des capteurs, de proposer des stratégies
(détection, traitement, risques associés) de traitement des valeurs manquantes, des
doublons, des valeurs aberrantes, des variables catégorielles

prompt :

Voici un dataset de capteurs IoT contenant des colonnes temperature, humidite, occupants, consommation, avec des valeurs manquantes, des doublons potentiels, des valeurs aberrantes et des variables catégorielles.
Propose, pour chacun de ces problèmes (valeurs manquantes, doublons, valeurs aberrantes, variables catégorielles) :
- une méthode de détection
- une méthode de traitement
- les risques associés si le problème n'est pas traité

Réponse du LLM :

1. Valeurs manquantes

Constat dans le fichier : température (6), humidité (5), pression (5), consommation (5), état (4).

Détection : df.isna().sum() pour compter par colonne ; df.isna().mean()*100 pour le taux ; une heatmap (sns.heatmap(df.isna())) pour visualiser leur répartition (aléatoire ou concentrée sur certains capteurs/dates).

Traitement : pour les variables numériques (temp, humidité, pression, consommation), imputation par la médiane ou par interpolation temporelle (df.interpolate()) puisqu'il s'agit de séries de capteurs dans le temps ; pour etat (catégorielle), imputation par le mode ou par une catégorie "Inconnu" plutôt que de deviner un état.

Risques si non traité : erreurs ou plantages lors des calculs (moyennes, modèles ML) ; biais si les valeurs manquantes ne sont pas aléatoires (ex : un capteur défaillant qui ne remonte plus de données à certaines heures) ; sous-estimation de la variabilité réelle.

4) Créer un prompt permettant d’expliquer les métriques de classification (Accuracy, Precision,
Recall, F1-score, ROC-AUC). Pour chacune, le LLM doit donner la définition, expliquer son
interprétation, donner un exemple concret et indiquer dans quel contexte elle est
particulièrement utile.

prompt:

Explique les métriques de classification suivantes : Accuracy, Precision, Recall, F1-score, ROC-AUC.
Pour chacune, donne : la définition, l'interprétation, un exemple concret, et le contexte dans lequel elle est particulièrement utile.

Réponse du LLM:

Métriques de classification
1. Accuracy (exactitude)
Définition : proportion de prédictions correctes sur l'ensemble des prédictions : (VP + VN) / (VP + VN + FP + FN)
Interprétation : une accuracy de 0,90 signifie que le modèle a correctement classé 90% des observations
Exemple concret : sur 100 emails classés comme spam/non-spam, si le modèle en classe correctement 92, l'accuracy est de 92%

5) Créer un prompt permettant d’expliquer les métriques de Régression (MAE, MSE, RMSE).

exemple concret et indiquer dans quel contexte elle est particulièrement utile.

prompt:

Explique les métriques de régression suivantes : MAE, MSE, RMSE.
Pour chacune, donne : la définition, l'interprétation, un exemple concret, et le contexte dans lequel elle est particulièrement utile.

Réponse du LLM:

Métriques de régression
1. MAE (Mean Absolute Error / Erreur absolue moyenne)
Définition : moyenne des valeurs absolues des écarts entre valeurs prédites et valeurs réelles : MAE = (1/n) × Σ|y_réel - y_prédit|
Interprétation : exprimée dans la même unité que la variable cible, elle indique l'erreur moyenne "typique" du modèle, sans distinguer si l'erreur vient d'une sur- ou sous-estimation
